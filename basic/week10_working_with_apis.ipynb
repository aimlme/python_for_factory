## 📌 **1. Introduction to APIs**

- **API (Application Programming Interface):** A set of rules for communication between applications.
- APIs allow you to **fetch data**, **send data**, and interact with external services.

### 📝 Example of APIs:
- Weather API (e.g., OpenWeatherMap)
- Currency API (e.g., ExchangeRate API)
- Social Media APIs (e.g., Twitter, Facebook)

APIs typically return data in **JSON format**.

Let's learn how to interact with APIs using Python!



## 📌 **2. Making API Requests with `requests`**

The `requests` library in Python is used to make HTTP requests to APIs.

### 📝 Example of Making an API Request:


````python
import requests

# Example API Request
response = requests.get("https://api.github.com")
print("Status Code:", response.status_code)
print("Response Headers:", response.headers)
````


## 📌 **3. JSON Parsing**

Most APIs return data in **JSON (JavaScript Object Notation)** format.

- Use `response.json()` to parse JSON data.

### 📝 Example of Parsing JSON:


````python
import requests

# Example JSON Parsing
response = requests.get("https://api.github.com")
data = response.json()

# Print keys from JSON data
print("Keys in Response JSON:", data.keys())

# Access specific data
print("API URL:", data['current_user_url'])
````


## 🚀 **4. Hands-On Project: Create a Currency Converter with an API**

### 🎯 **Project Goals:**
1. Fetch **real-time currency exchange rates** using an API.
2. Allow users to **convert amounts** between currencies.

### 📝 Prerequisites:
- Install `requests` library (if not already installed):
```bash
pip install requests
```
- Get a free API key from [ExchangeRate-API](https://www.exchangerate-api.com/).

### 📝 Example Code:


````python
import requests

# Currency Converter Example
API_KEY = "your_api_key"  # Replace with your API key
BASE_URL = "https://v6.exchangerate-api.com/v6/"

def get_exchange_rate(base_currency, target_currency):
    try:
        url = f"{BASE_URL}{API_KEY}/latest/{base_currency}"
        response = requests.get(url)
        data = response.json()
        
        if response.status_code == 200:
            rate = data['conversion_rates'][target_currency]
            return rate
        else:
            return f"Error: {data['error-type']}"
    except Exception as e:
        return f"Error: {e}"

# User input
base_currency = input("Enter base currency (e.g., USD): ").upper()
target_currency = input("Enter target currency (e.g., EUR): ").upper()
amount = float(input("Enter amount to convert: "))

rate = get_exchange_rate(base_currency, target_currency)
if isinstance(rate, float):
    converted_amount = amount * rate
    print(f"{amount} {base_currency} = {converted_amount:.2f} {target_currency}")
else:
    print(rate)
````
