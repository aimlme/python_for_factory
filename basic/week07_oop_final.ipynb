## 📌 1. What are Classes and Objects?


- **Class** = A blueprint (plan) for creating objects.
- **Object** = A real-world thing created from a class.

For example, a **Car class** describes all cars, but **my_car** is an actual car.


```python

# Example: Creating a Person class and making an object

class Person:
    def __init__(self, name, age):  # This is a constructor (runs automatically)
        self.name = name
        self.age = age

    def introduce(self):  # A function inside a class is called a method
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

# Creating an object (a real person)
person1 = Person("Alice", 30)
person1.introduce()
```

## 📌 2. What is a Constructor (`__init__`)?


- The `__init__` method runs **automatically** when an object is created.
- It helps **set up** an object with initial values.


```python

class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def show_info(self):
        print(f"Car: {self.brand} {self.model}")

# Creating an object (a real car)
my_car = Car("Toyota", "Corolla")
my_car.show_info()
```

## 📌 3. What is Encapsulation? (Hiding Important Data)


- **Encapsulation** means **hiding** data so it cannot be changed directly.
- This keeps the object **safe** from accidental changes.


```python

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private variable (cannot be accessed directly)

    def deposit(self, amount):
        self.__balance += amount
        print(f"Deposited {amount}. New balance: {self.__balance}")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            print(f"Withdrew {amount}. Remaining balance: {self.__balance}")

    def show_balance(self):
        print(f"Balance: {self.__balance}")

# Creating an account
account = BankAccount("Alice", 1000)
account.deposit(500)  # Depositing money
account.withdraw(300) # Withdrawing money
account.show_balance() # Checking balance
```

## 📌 4. What is Inheritance? (Reusing Code)


- **Inheritance** allows one class to take features from another class.
- It **saves time** because we don’t have to write the same code again.


```python

# Parent class (Animal)
class Animal:
    def __init__(self, name):
        self.name = name  # Every animal has a name

    def make_sound(self):
        print("Some generic animal sound")

# Child class (Dog inherits from Animal)
class Dog(Animal):
    def make_sound(self):
        print(f"{self.name} says: Woof! Woof!")

# Child class (Cat inherits from Animal)
class Cat(Animal):
    def make_sound(self):
        print(f"{self.name} says: Meow!")

# Creating objects
dog = Dog("Buddy")
cat = Cat("Whiskers")

dog.make_sound()  # Buddy says: Woof! Woof!
cat.make_sound()  # Whiskers says: Meow!
```

## 📌 5. What is Polymorphism? (Same Function, Different Behavior)


- **Polymorphism** means that different classes can **use the same function** in different ways.
- Example: Different animals make different sounds, but they all have a `make_sound()` function.


```python

class Animal:
    def make_sound(self):
        print("Some animal sound")

class Dog(Animal):
    def make_sound(self):
        print("Woof! Woof!")

class Cat(Animal):
    def make_sound(self):
        print("Meow!")

# Using polymorphism
animals = [Dog(), Cat(), Animal()]

for animal in animals:
    animal.make_sound()  # Woof! Woof!  Meow!  Some animal sound
```

## 📌 6. Hands-on Exercise: Build a Simple Library System


Let’s apply everything we learned by creating a **Library System**.

### **Requirements:**
- We need a **Book class** with a title and an author.
- We need a **borrow() method** to mark a book as borrowed.
- We need a **Library class** to manage multiple books.

### **Expected Behavior:**
- A library has many books.
- We can borrow books.
- We can check which books are available.


```python

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_borrowed = False  # All books start as available

    def borrow(self):
        if self.is_borrowed:
            print(f"{self.title} is already borrowed.")
        else:
            self.is_borrowed = True
            print(f"You have borrowed {self.title}.")

# Creating books
book1 = Book("1984", "George Orwell")
book2 = Book("Brave New World", "Aldous Huxley")

# Borrowing books
book1.borrow()  # You have borrowed 1984
book1.borrow()  # 1984 is already borrowed
```

```python
import tkinter as tk

# Function to update the expression in the entry field
def press(num):
    entry_text.set(entry_text.get() + str(num))

# Function to evaluate the final expression
def equal_press():
    try:
        total = str(eval(entry_text.get()))
        entry_text.set(total)
    except:
        entry_text.set("Error")

# Function to clear the entry field
def clear():
    entry_text.set("")

# Create main window
root = tk.Tk()
root.title("Calculator")
root.geometry("300x400")
root.resizable(False, False)

# Entry field for calculator
entry_text = tk.StringVar()
entry = tk.Entry(root, textvariable=entry_text, font=("Arial", 20), bd=10, relief="ridge", justify="right")
entry.grid(row=0, column=0, columnspan=4, ipadx=8, ipady=10)

# Buttons layout
buttons = [
    ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
    ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
    ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
    ('0', 4, 0), ('.', 4, 1), ('+', 4, 2), ('=', 4, 3),
    ('C', 5, 0)
]

# Adding buttons to the window
for (text, row, col) in buttons:
    if text == "=":
        tk.Button(root, text=text, width=10, height=2, bg="lightblue", font=("Arial", 14),
                  command=equal_press).grid(row=row, column=col, columnspan=2, sticky="nsew")
    elif text == "C":
        tk.Button(root, text=text, width=10, height=2, bg="red", fg="white", font=("Arial", 14),
                  command=clear).grid(row=row, column=col, columnspan=4, sticky="nsew")
    else:
        tk.Button(root, text=text, width=5, height=2, font=("Arial", 14),
                  command=lambda t=text: press(t)).grid(row=row, column=col, sticky="nsew")

# Run the Tkinter event loop
root.mainloop()
```

```python
import tkinter as tk

class Calculator(tk.Tk):
    def __init__(self):
        super().__init__()

        self.title("Calculator")
        self.geometry("300x400")

        self.entry_text = tk.StringVar()
        self.entry = tk.Entry(self, textvariable=self.entry_text, font=("Arial", 20), justify="right")
        self.entry.grid(row=0, column=0, columnspan=4)

        self.create_buttons()

    def create_buttons(self):
        buttons = [
            ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
            ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
            ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
            ('0', 4, 0), ('.', 4, 1), ('+', 4, 2), ('=', 4, 3),
            ('C', 5, 0)
        ]

        for (text, row, col) in buttons:
            if text == "=":
                tk.Button(self, text=text, width=10, height=2, command=self.equal_press).grid(row=row, column=col, columnspan=2)
            elif text == "C":
                tk.Button(self, text=text, width=10, height=2, command=self.clear).grid(row=row, column=col, columnspan=4)
            else:
                tk.Button(self, text=text, width=5, height=2, command=lambda t=text: self.press(t)).grid(row=row, column=col)

    def press(self, num):
        self.entry_text.set(self.entry_text.get() + str(num))

    def equal_press(self):
        try:
            self.entry_text.set(str(eval(self.entry_text.get())))
        except:
            self.entry_text.set("Error")

    def clear(self):
        self.entry_text.set("")

if __name__ == "__main__":
    app = Calculator()
    app.mainloop()
```
