## 📌 **1. Anatomy of a Matplotlib Figure**

### 📝 Key Concepts:
- **Figure** — the whole window/canvas.
- **Axes** — an individual plot (a figure can contain multiple axes/subplots).
- **Axis** — the x-axis / y-axis of an Axes, with ticks and labels.

### 📝 Install & Import

```python
# Install if needed (uncomment)
# !pip install matplotlib

import matplotlib.pyplot as plt
import numpy as np
print("Matplotlib loaded")
```

## 📌 **2. Basic Line Plot**

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y, marker='o')
plt.title('Line Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

## 📌 **3. Customizing Plots**

Color, line style, markers, legend, and grid all help make a chart clearer.

```python
x = np.linspace(0, 10, 50)
y = np.sin(x)

plt.plot(x, y, color='crimson', linestyle='--', linewidth=2, marker='.', label='sin(x)')
plt.title('Customized Line Plot')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 📌 **4. Multiple Lines on One Chart**

```python
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.title('Multiple Lines')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

## 📌 **5. Subplots**

Use `plt.subplots()` to place multiple charts in one figure.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

x = np.linspace(0, 10, 100)
axes[0].plot(x, np.sin(x), color='steelblue')
axes[0].set_title('Sine')

axes[1].plot(x, np.cos(x), color='darkorange')
axes[1].set_title('Cosine')

plt.tight_layout()
plt.show()
```

```python
# A 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(8, 6))
data_sets = [np.random.default_rng(i).normal(0, 1, 200) for i in range(4)]

for ax, data, i in zip(axes.flat, data_sets, range(4)):
    ax.hist(data, bins=15, color='teal', alpha=0.7)
    ax.set_title(f'Dataset {i+1}')

plt.tight_layout()
plt.show()
```

## 📌 **6. Bar Charts**

```python
categories = ['Line A', 'Line B', 'Line C', 'Line D']
output = [420, 380, 510, 460]

plt.bar(categories, output, color='mediumseagreen')
plt.title('Daily Output by Production Line')
plt.xlabel('Line')
plt.ylabel('Units Produced')
plt.show()
```

```python
# Horizontal bar chart
plt.barh(categories, output, color='cornflowerblue')
plt.title('Daily Output by Production Line (Horizontal)')
plt.xlabel('Units Produced')
plt.show()
```

## 📌 **7. Scatter Plots**

```python
rng = np.random.default_rng(7)
cycle_time = rng.normal(12, 1.5, 60)
defect_count = (cycle_time - 12) * 2 + rng.normal(0, 1, 60)

plt.scatter(cycle_time, defect_count, color='indianred', alpha=0.7)
plt.title('Cycle Time vs Defect Count')
plt.xlabel('Cycle Time (s)')
plt.ylabel('Defect Count')
plt.grid(True, alpha=0.3)
plt.show()
```

## 📌 **8. Histograms**

Histograms show the **distribution** of a numeric variable.

```python
rng = np.random.default_rng(3)
measurements = rng.normal(loc=50, scale=2, size=500)  # e.g. part thickness in mm

plt.hist(measurements, bins=25, color='slateblue', edgecolor='white')
plt.title('Distribution of Part Thickness')
plt.xlabel('Thickness (mm)')
plt.ylabel('Frequency')
plt.show()
```

## 📌 **9. Pie Charts**

```python
labels = ['Scratches', 'Dents', 'Misalignment', 'Other']
sizes = [45, 25, 20, 10]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Defect Type Breakdown')
plt.axis('equal')  # keep the pie circular
plt.show()
```

## 📌 **10. Box Plots**

Box plots are great for comparing distributions and spotting outliers across groups.

```python
rng = np.random.default_rng(9)
line_a = rng.normal(12, 1.0, 100)
line_b = rng.normal(13, 1.8, 100)
line_c = rng.normal(11.5, 0.7, 100)

plt.boxplot([line_a, line_b, line_c], labels=['Line A', 'Line B', 'Line C'])
plt.title('Cycle Time Distribution by Line')
plt.ylabel('Cycle Time (s)')
plt.show()
```

## 📌 **11. Saving Figures to Files**

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title('Sine Wave')

plt.savefig('sine_wave.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved to sine_wave.png")
```

## 📌 **12. Styling with `plt.style`**

```python
print("Available styles (first 8):", plt.style.available[:8])

with plt.style.context('ggplot'):
    x = np.linspace(0, 10, 100)
    plt.plot(x, np.sin(x), label='sin(x)')
    plt.plot(x, np.cos(x), label='cos(x)')
    plt.title('Styled with ggplot theme')
    plt.legend()
    plt.show()
```

## 🏭 **13. Hands-On Exercise: Visualizing Production KPIs**

Using the same style of data as the Pandas supplement, let's visualize **daily output vs defect rate** for a production line.

### 🎯 Goals:
1. Plot daily output as a bar chart.
2. Overlay defect rate as a line on a secondary y-axis.
3. Combine both into a single, readable chart.

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
output = [1200, 1100, 1250, 1050, 1300, 1150]
defect_rate = [2.0, 3.0, 1.6, 4.0, 1.4, 3.0]

fig, ax1 = plt.subplots(figsize=(8, 5))

ax1.bar(days, output, color='skyblue', label='Output (units)')
ax1.set_xlabel('Day')
ax1.set_ylabel('Output (units)', color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

ax2 = ax1.twinx()  # secondary y-axis sharing the same x-axis
ax2.plot(days, defect_rate, color='crimson', marker='o', linewidth=2, label='Defect Rate (%)')
ax2.set_ylabel('Defect Rate (%)', color='crimson')
ax2.tick_params(axis='y', labelcolor='crimson')

plt.title('Daily Output vs Defect Rate')
fig.tight_layout()
plt.show()
```
