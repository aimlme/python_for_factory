## 📌 **1. Series vs DataFrame**

### 📝 Key Concepts:
- **Series** = a single labeled 1D column of data.
- **DataFrame** = a 2D table made of multiple Series (columns), like a spreadsheet.

### 📝 Install & Import

```python
# Install if needed (uncomment)
# !pip install pandas

import pandas as pd
print("Pandas version:", pd.__version__)
```

```python
# A Series
s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])
print("Series:\n", s)
print("\nValue at index 'b':", s["b"])
```

## 📌 **2. Creating DataFrames**

```python
# From a dictionary of lists
data = {
    "Name": ["Alice", "Bob", "Charlie", "Dana"],
    "Age": [25, 30, 35, 28],
    "Department": ["QA", "Production", "Maintenance", "QA"],
    "Salary": [50000, 60000, 55000, 52000],
}
df = pd.DataFrame(data)
df
```

```python
# From a list of dictionaries (rows)
rows = [
    {"Station": "ST01", "Cycle_Time": 12.5},
    {"Station": "ST02", "Cycle_Time": 11.8},
    {"Station": "ST03", "Cycle_Time": 13.2},
]
df_rows = pd.DataFrame(rows)
df_rows
```

## 📌 **3. Reading and Writing Files**

Pandas can read/write CSV, Excel, JSON, SQL and more.

```python
# Write our DataFrame to CSV
df.to_csv("employees.csv", index=False)

# Read it back
df_from_csv = pd.read_csv("employees.csv")
print(df_from_csv)

# Writing/reading Excel requires openpyxl:
# df.to_excel("employees.xlsx", index=False)
# df_from_excel = pd.read_excel("employees.xlsx")
```

## 📌 **4. Inspecting Data**

```python
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nData types:\n", df.dtypes)
print("\nFirst 2 rows:\n", df.head(2))
print("\nSummary statistics:\n", df.describe())
```

## 📌 **5. Selecting and Filtering Data**

### 📝 Selecting columns

```python
print(df["Name"])              # single column -> Series
print(df[["Name", "Salary"]])  # multiple columns -> DataFrame
```

### 📝 `loc` (label-based) vs `iloc` (position-based)

```python
print("loc (row label 1):\n", df.loc[1])
print("\niloc (first row by position):\n", df.iloc[0])
print("\nloc slice with column names:\n", df.loc[0:2, ["Name", "Department"]])
```

### 📝 Boolean filtering

```python
high_earners = df[df["Salary"] > 52000]
print("Employees earning more than 52,000:\n", high_earners)

qa_dept = df[(df["Department"] == "QA") & (df["Age"] < 30)]
print("\nQA employees younger than 30:\n", qa_dept)
```

## 📌 **6. Handling Missing Data**

```python
import numpy as np

df_missing = pd.DataFrame({
    "Machine": ["M1", "M2", "M3", "M4"],
    "Output": [120, np.nan, 135, np.nan],
    "Defects": [3, 2, np.nan, 1],
})
print("Original:\n", df_missing)

print("\nIs null:\n", df_missing.isna())
print("\nDrop rows with any NaN:\n", df_missing.dropna())
print("\nFill NaN with 0:\n", df_missing.fillna(0))
print("\nFill NaN with column mean:\n", df_missing.fillna(df_missing.mean(numeric_only=True)))
```

## 📌 **7. Adding, Modifying, and Dropping Columns**

```python
df2 = df.copy()

# Add a new column
df2["Bonus"] = df2["Salary"] * 0.10

# Modify a column using a function
df2["Name_Upper"] = df2["Name"].apply(lambda x: x.upper())

# Conditional column with np.where
df2["Senior"] = np.where(df2["Age"] >= 30, "Yes", "No")

print(df2)

# Drop a column
df2 = df2.drop(columns=["Name_Upper"])
print("\nAfter dropping Name_Upper:\n", df2.columns.tolist())
```

## 📌 **8. Sorting Data**

```python
print("Sorted by Salary (descending):\n", df.sort_values("Salary", ascending=False))
print("\nSorted by Department then Age:\n", df.sort_values(["Department", "Age"]))
```

## 📌 **9. Grouping and Aggregation**

```python
grouped = df.groupby("Department")["Salary"].mean()
print("Average salary per department:\n", grouped)

agg = df.groupby("Department").agg(
    avg_salary=("Salary", "mean"),
    headcount=("Name", "count"),
    max_age=("Age", "max"),
)
print("\nMulti-aggregation:\n", agg)
```

```python
# Pivot table example
sales = pd.DataFrame({
    "Line": ["A", "A", "B", "B", "A", "B"],
    "Shift": ["Day", "Night", "Day", "Night", "Day", "Day"],
    "Units": [100, 90, 110, 95, 105, 120],
})
pivot = sales.pivot_table(values="Units", index="Line", columns="Shift", aggfunc="sum")
print(pivot)
```

## 📌 **10. Merging, Joining, and Concatenating**

```python
employees = pd.DataFrame({"EmpID": [1, 2, 3], "Name": ["Alice", "Bob", "Charlie"]})
salaries = pd.DataFrame({"EmpID": [1, 2, 4], "Salary": [50000, 60000, 45000]})

# Inner join (only matching EmpIDs)
inner = pd.merge(employees, salaries, on="EmpID", how="inner")
print("Inner join:\n", inner)

# Left join (keep all employees)
left = pd.merge(employees, salaries, on="EmpID", how="left")
print("\nLeft join:\n", left)

# Concatenate two DataFrames vertically
more_employees = pd.DataFrame({"EmpID": [5], "Name": ["Dana"]})
combined = pd.concat([employees, more_employees], ignore_index=True)
print("\nConcatenated:\n", combined)
```

## 📌 **11. Working with Dates and Times**

```python
dates = pd.DataFrame({
    "Date": ["2026-01-01", "2026-01-02", "2026-01-03"],
    "Output": [500, 480, 510],
})
dates["Date"] = pd.to_datetime(dates["Date"])
dates = dates.set_index("Date")
print(dates)
print("\nDay of week:\n", dates.index.day_name())

# Resample to a 2-day sum (example of time-based aggregation)
print("\n2-day resample sum:\n", dates.resample("2D").sum())
```

## 🏭 **12. Hands-On Exercise: Production KPI Data**

A production line logs daily **output** and **defect count** per shift. We'll use Pandas to calculate a simple **defect rate KPI**.

### 🎯 Goals:
1. Build a small production dataset.
2. Compute defect rate per row.
3. Find the shift with the worst average defect rate.

```python
production = pd.DataFrame({
    "Date": pd.date_range("2026-08-01", periods=6, freq="D"),
    "Shift": ["Day", "Night", "Day", "Night", "Day", "Night"],
    "Output": [1200, 1100, 1250, 1050, 1300, 1150],
    "Defects": [24, 33, 20, 42, 18, 35],
})

production["Defect_Rate_%"] = (production["Defects"] / production["Output"] * 100).round(2)
print(production)

worst_shift = production.groupby("Shift")["Defect_Rate_%"].mean().idxmax()
print(f"\nShift with the highest average defect rate: {worst_shift}")

print("\nSummary by shift:\n", production.groupby("Shift")["Defect_Rate_%"].mean().round(2))
```
