## 📌 **1. Variables and Constants**

- **Variable:** A container for storing data values.  
- **Constant:** A value that should not change during program execution (by convention, use uppercase names).

### 📝 Example of Variables and Constants:


```python
# Variables
name = "Alice"
age = 25

# Constants (by convention)
PI = 3.14159

# Print values
print("Name:", name)
print("Age:", age)
print("Value of PI:", PI)
```


## 📌 **2. Data Types**

Python supports several built-in data types:

- **int:** Integer numbers
- **float:** Floating-point numbers
- **str:** Strings (text)
- **bool:** Boolean (True/False)
- **list:** Ordered, mutable collections
- **tuple:** Ordered, immutable collections
- **dict:** Key-value pairs
- **set:** Unordered, unique items

### 📝 Example of Different Data Types:


```python
# Examples of Data Types
integer_var = 10
float_var = 3.14
string_var = "Hello, World"
bool_var = True
list_var = [1, 2, 3]
tuple_var = (1, 2, 3)
dict_var = {"key": "value"}
set_var = {1, 2, 3}

# Print Types
print(type(integer_var))
print(type(float_var))
print(type(string_var))
print(type(bool_var))
print(type(list_var))
print(type(tuple_var))
print(type(dict_var))
print(type(set_var))
```


## 📌 **3. Type Casting**

- **Implicit Casting:** Python automatically converts data types.
- **Explicit Casting:** You manually convert data types using functions like `int()`, `float()`, `str()`.

### 📝 Example of Type Casting:


```python
# Explicit Type Casting
x = "5"
y = int(x)  # Convert string to integer

a = 3.14
b = int(a)  # Convert float to integer

# Print results
print(y, type(y))
print(b, type(b))
```


## 📌 **4. Input/Output**

- **`print()`**: Display output on the screen.
- **`input()`**: Get user input.

### 📝 Example of Input/Output:


```python
# Input Example
name = input("Enter your name: ")
age = int(input("Enter your age: "))

# Output Example
print(f"Hello {name}, you are {age} years old.")
```


## 🚀 **5. Hands-On Exercise: Build a Simple Calculator**

Let's build a simple calculator that:
1. Asks the user for two numbers.
2. Performs basic operations: `+`, `-`, `*`, `/`.

### 📝 Example of a Simple Calculator:


```python
# Simple Calculator
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Select an operation:")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (*)")
print("4. Division (/)")

operation = input("Enter the operation (+, -, *, /): ")

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error: Division by zero!"
else:
    result = "Invalid operation!"

print("Result:", result)
```
