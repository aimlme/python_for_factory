## 📌 **1. NumPy for Numerical Data**

**NumPy (Numerical Python)** is a library for numerical computations in Python.

### 📝 Key Features of NumPy:
- Support for **multi-dimensional arrays**.
- Mathematical and logical operations on arrays.
- Linear algebra, Fourier analysis, and more.

### 📝 Example:


````python
import numpy as np

# Create a NumPy array
array = np.array([1, 2, 3, 4, 5])
print("Array:", array)

# Perform mathematical operations
print("Sum:", np.sum(array))
print("Mean:", np.mean(array))
print("Standard Deviation:", np.std(array))
````


## 📌 **2. Pandas for Data Manipulation**

**Pandas** is used for data manipulation and analysis.

### 📝 Key Features of Pandas:
- Data structures: **Series** and **DataFrame**.
- Handling missing data.
- Data filtering, grouping, and aggregation.

### 📝 Example:


````python
import pandas as pd

# Create a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35], 'Salary': [50000, 60000, 70000]}
df = pd.DataFrame(data)

# Display DataFrame
print(df)

# Filter data
print("Employees with Salary > 55000:")
print(df[df['Salary'] > 55000])
````


## 📌 **3. Matplotlib for Data Visualization**

**Matplotlib** is used for creating static, animated, and interactive visualizations.

### 📝 Common Plot Types:
- Line Plot
- Bar Chart
- Scatter Plot

### 📝 Example:


````python
import matplotlib.pyplot as plt

# Line Plot
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y, marker='o')
plt.title('Line Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
````


## 🚀 **4. Hands-On Exercise: Analyze a Dataset**

### 🎯 **Project Goals:**
1. Load a sample dataset.
2. Perform data manipulation using **Pandas**.
3. Visualize insights using **Matplotlib**.

### 📝 Prerequisites:
- Install libraries if not already installed:
```bash
pip install numpy pandas matplotlib
```

### 📝 Example Code:


````python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset (Example: Iris dataset from seaborn library)
import seaborn as sns
data = sns.load_dataset('iris')

# Display first 5 rows
print(data.head())

# Perform data analysis
avg_petal_length = data.groupby('species')['petal_length'].mean()
print(avg_petal_length)

# Visualize the results
avg_petal_length.plot(kind='bar', color='skyblue')
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.show()
````
