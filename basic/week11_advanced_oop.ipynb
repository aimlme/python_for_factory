## 📌 **1. Property Decorators (`@property`)**

- **Property Decorators** allow you to define methods that can be accessed like **attributes**.
- Common decorators include:
   - `@property` (getter)
   - `@<property>.setter` (setter)
   - `@<property>.deleter` (deleter)

### 📝 Example:


````python
# Example of Property Decorators

class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        self._name = value

    @property
    def age(self):
        return self._age

person = Person("Alice", 25)
print(person.name)  # Access property
person.name = "Bob"  # Modify property
print(person.name)
````


## 📌 **2. `super()` and Method Resolution Order (MRO)**

- `super()` is used to call **parent class methods**.
- **Method Resolution Order (MRO):** The order in which methods are inherited from classes.

### 📝 Check MRO:
```python
ClassName.__mro__
```

### 📝 Example:


````python
# Example of super() and MRO

class Parent:
    def show(self):
        print("Parent class method")

class Child(Parent):
    def show(self):
        super().show()
        print("Child class method")

# Test MRO
child = Child()
child.show()
print(Child.__mro__)
````


## 📌 **3. Multiple Inheritance**

- Multiple inheritance allows a class to **inherit from more than one class**.
- Python uses **C3 Linearization (MRO)** to resolve conflicts.

### 📝 Example:


````python
# Example of Multiple Inheritance

class A:
    def method(self):
        print("Class A Method")

class B:
    def method(self):
        print("Class B Method")

class C(A, B):
    def method(self):
        super().method()
        print("Class C Method")

# Test Multiple Inheritance
obj = C()
obj.method()
print(C.__mro__)
````


## 🚀 **4. Hands-On Exercise: Build an Employee Management System**

### 🎯 **Project Goals:**
1. Create an **Employee** class with attributes (`name`, `id`, `salary`).
2. Implement **@property** for salary to ensure valid salary updates.
3. Use **inheritance** for different employee types (e.g., Manager, Developer).
4. Display employee details.

### 📝 Example Code:


````python
# Employee Management System

class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self._salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value > 0:
            self._salary = value
        else:
            raise ValueError("Salary must be positive!")

    def display_info(self):
        print(f"Name: {self.name}, ID: {self.emp_id}, Salary: {self.salary}")

class Manager(Employee):
    def __init__(self, name, emp_id, salary, team_size):
        super().__init__(name, emp_id, salary)
        self.team_size = team_size

    def display_info(self):
        super().display_info()
        print(f"Team Size: {self.team_size}")

class Developer(Employee):
    def __init__(self, name, emp_id, salary, programming_language):
        super().__init__(name, emp_id, salary)
        self.programming_language = programming_language

    def display_info(self):
        super().display_info()
        print(f"Programming Language: {self.programming_language}")

# Test Employee System
emp1 = Manager("Alice", 101, 8000, 5)
emp2 = Developer("Bob", 102, 6000, "Python")

emp1.display_info()
emp2.display_info()

# Update salary
emp1.salary = 9000
print(f"Updated Salary: {emp1.salary}")
````
