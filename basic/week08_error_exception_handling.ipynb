## 📌 **1. Types of Exceptions**

Exceptions are **errors** detected during execution. Python provides **built-in exception classes**.

### 📝 Common Exception Types:
- `ValueError`: Invalid argument.
- `TypeError`: Invalid data type.
- `IndexError`: Index out of range.
- `KeyError`: Key not found in a dictionary.
- `ZeroDivisionError`: Division by zero.

### 📝 Example:


````python
# Example of Exceptions

try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print("Result:", result)
except ValueError:
    print("Invalid input! Please enter a number.")
except ZeroDivisionError:
    print("Cannot divide by zero!")
except Exception as e:
    print(f"Unexpected error: {e}")
````


## 📌 **2. Custom Exception Classes**

You can create **custom exception classes** for specific scenarios.

### 📝 Example:


````python
# Custom Exception Class

class CustomError(Exception):
    def __init__(self, message):
        super().__init__(message)

try:
    raise CustomError("This is a custom exception!")
except CustomError as e:
    print(e)
````


## 📌 **3. Debugging Techniques**

Debugging helps identify and fix errors. Common techniques include:
- **print statements:** Print variable values to understand program flow.
- **Using a Debugger:** Step through code using an IDE debugger.
- **`pdb` module:** Python's built-in debugger.

### 📝 Example with `pdb`:
Run the following code in your terminal:
```python
import pdb

x = 5
y = 10
pdb.set_trace()
result = x + y
print(result)
```
You can use commands like `n` (next), `c` (continue), and `q` (quit) in the debugger.



## 📌 **4. Logging**

Logging allows you to **record events** and **errors** during program execution.

### 📝 Logging Levels:
- `DEBUG`: Detailed information for diagnosing problems.
- `INFO`: Confirm program is working.
- `WARNING`: Indicate a potential problem.
- `ERROR`: Indicate a serious issue.
- `CRITICAL`: Indicate a severe error.

### 📝 Example:


````python
import logging

# Configure logging
logging.basicConfig(level=logging.DEBUG, filename='app.log', filemode='w',
                    format='%(name)s - %(levelname)s - %(message)s')

try:
    x = 10 / 0
except ZeroDivisionError:
    logging.error("Division by zero occurred!")
````


## 🚀 **5. Hands-On Exercise: Create a Robust Login System**

### 🎯 **Project Goals:**
1. Validate **username** and **password**.
2. Use **exception handling** for errors.
3. Log every **successful** and **failed login attempt**.

### 📝 Example Code:


````python
import logging

# Configure logging
logging.basicConfig(level=logging.INFO, filename='login.log', filemode='w',
                    format='%(asctime)s - %(levelname)s - %(message)s')

# Predefined credentials
USERNAME = "admin"
PASSWORD = "password123"

class LoginError(Exception):
    pass

def login(username, password):
    if username != USERNAME:
        logging.warning("Failed login attempt: Invalid username")
        raise LoginError("Invalid Username")
    if password != PASSWORD:
        logging.warning("Failed login attempt: Invalid password")
        raise LoginError("Invalid Password")
    logging.info("Successful login")
    return "Login successful!"

# Test Login
try:
    user = input("Enter username: ")
    pwd = input("Enter password: ")
    print(login(user, pwd))
except LoginError as e:
    print(f"Login failed: {e}")
except Exception as e:
    logging.critical(f"Unexpected error: {e}")
````
