## 📌 **1. Understanding Python Modules and Packages**

- **Module:** A Python file containing functions, classes, and variables (`.py` file).  
- **Package:** A directory containing multiple modules and a special `__init__.py` file.

### 📝 Example of a Module:
Create a file named `mymodule.py`:
```python
def greet(name):
    return f"Hello, {name}!"
```
Then import it:
```python
import mymodule
print(mymodule.greet("Alice"))
```



## 📌 **2. Installing Packages with `pip`**

- `pip` is Python's package manager.
- Use it to **install**, **upgrade**, and **uninstall** packages.

### 📝 Common pip Commands:
- `pip install package_name`
- `pip uninstall package_name`
- `pip list` (List installed packages)
- `pip freeze > requirements.txt` (Export dependencies)
- `pip install -r requirements.txt` (Install dependencies)

### 📝 Example:
Run the following in your terminal:
```bash
pip install requests
```
Then use the installed package in Python:
```python
import requests
response = requests.get("https://api.github.com")
print(response.status_code)
```



## 📌 **3. Virtual Environments**

- A **virtual environment** isolates dependencies for different projects.
- Prevents conflicts between packages.

### 📝 Create a Virtual Environment:
```bash
python -m venv myenv
```

### 📝 Activate the Virtual Environment:
- **Windows:** `myenv\Scripts\activate`
- **Mac/Linux:** `source myenv/bin/activate`

### 📝 Deactivate Virtual Environment:
```bash
deactivate
```



## 🚀 **4. Hands-On Exercise: Build a Weather App using an API**

### 🎯 **Project Goals:**
1. Fetch weather data using an **API** (e.g., OpenWeatherMap API).
2. Display **temperature**, **humidity**, and **weather description**.

### 📝 Prerequisites:
- Install `requests` library:
```bash
pip install requests
```
- Get an **API Key** from [OpenWeatherMap](https://openweathermap.org/api).

### 📝 Example Code:


````python
import requests

# Weather App Example
API_KEY = "your_api_key"  # Replace with your actual API key
BASE_URL = "http://api.openweathermap.org/data/2.5/weather"

def get_weather(city):
    try:
        params = {"q": city, "appid": API_KEY, "units": "metric"}
        response = requests.get(BASE_URL, params=params)
        data = response.json()
        
        if response.status_code == 200:
            weather = {
                "City": data["name"],
                "Temperature": data["main"]["temp"],
                "Humidity": data["main"]["humidity"],
                "Description": data["weather"][0]["description"]
            }
            return weather
        else:
            return f"Error: {data['message']}"
    except Exception as e:
        return f"Error: {e}"

# Get city from user
city = input("Enter city name: ")
weather_info = get_weather(city)
print(weather_info)
````
